```python
students = [
    {
        "name": "Ram",
        "roll": 1,
        "marks": [75, 80, 70]
    },
    {
        "name": "Sita",
        "roll": 2,
        "marks": [85, 90, 88]
    }
]

# Add a student
def add_student(name, roll, marks):
    student = {
        "name": name,
        "roll": roll,
        "marks": marks
    }
    students.append(student)

# Display all students
def display_students():
    for student in students:
        print("Name:", student["name"])
        print("Roll Number:", student["roll"])
        print("Marks:", student["marks"])
        print()

# Calculate average mark of a student
def calculate_average(student):
    return sum(student["marks"]) / len(student["marks"])

# Find student with the highest average
def highest_average():
    highest_student = students[0]

    for student in students:
        if calculate_average(student) > calculate_average(highest_student):
            highest_student = student

    return highest_student


# Add a new student
add_student("Hari", 3, [78, 82, 75])


# Display all students
print("All Students:")
display_students()


# Display average marks
print("Average Marks:")
for student in students:
    average = calculate_average(student)
    print(student["name"], ":", average)


# Display student with highest average
top_student = highest_average()

print("\nStudent with Highest Average:")
print("Name:", top_student["name"])
print("Roll Number:", top_student["roll"])
print("Average:", calculate_average(top_student))
```
